```python
# Data Types

# Integer data type
x = 10
print("Integer:", x)    
y = -5
print("Negative Integer:", y)
z = 0
print("Zero Integer:", z)   
print('-'*30)

# Float data type
a = 3.14
print("Float:", a)
b = -2.5
print("Negative Float:", b)
c = 0.0
print("Zero Float:", c)
print('-'*30)

# String data type
name = "Alice"
print("String:", name)
greeting = "Hello, World!"
print("String:", greeting)
empty_string = ""
print("Empty String:", empty_string)
print('-'*30)

# Boolean data type
is_valid = True
print("Boolean (True):", is_valid)
is_false = False
print("Boolean (False):", is_false)
print('-'*30)

# Variables set 
age = 25
height = 1.75
is_student = True
name = "Bob"    
print("Integer_" + "Age:", age)
print("Float_" + "Height:", height)
print("Boolean_" + "Is Student:", is_student)
print("String_" + "Name:", name)
print('-'*30)

# Data types
x = 10
print("Integer Data type of x:",f"{x}", type(x))
y = 3.14    
print("Float Data type of y:", f"{y}", type(y))
name = "Alice"
print("String Data type of name:", f"{name}", type(name))
is_valid = True
print("Boolean Data type of is_valid:", f"{is_valid}", type(is_valid))
print('-'*30)
```
