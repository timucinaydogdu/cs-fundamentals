```python
# Lists
print("Empty list")
empty_list = []
print(empty_list)
print("-" * 30)

print("Item access")
my_list = [10, 20, 30, 40, 50]
print("my_list =", my_list)
print("First item:", my_list[0])
print("Last item:", my_list[-1])
print("-" * 30)

print("List Copying ")          # If you assign a list to a new variable, both variables point to the same list. 
                                # To create a copy of the list, you can use the copy() method.
original_list = [1, 2, 3]
copied_list = original_list.copy()
print("Original list:", original_list)
print("Copied list:", copied_list)
print("-" * 30)

# IOT List Sample
heat_data = [23.5, 24.0, 22.8, 25.1, 24.3]

last_temperature = heat_data[-3:]
print("Last three temperature readings:", last_temperature)

over_temperature_threshold = [x for x in heat_data if x > 24.0]
print("Temperatures over 24.0:", over_temperature_threshold)

average_temperature = sum(heat_data) / len(heat_data)
print("Average temperature:", average_temperature)

print("-" * 30)
```
