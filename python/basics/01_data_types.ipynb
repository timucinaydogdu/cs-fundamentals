```python
# Data maths operations
print("-" * 30)
print("Data maths operations")
print("Addition: 5 + 3 =", 5 + 3) 
print("Subtraction: 5 - 3 =", 5 - 3)
print("Multiplication: 5 * 3 =", 5 * 3)
print("Division: 5 / 3 =", 5 / 3)
print("Floor Division: 5 // 3 =", 5 // 3)
print("Modulus: 5 % 3 =", 5 % 3)
print("Exponentiation: 5 ** 3 =", 5 ** 3)
print("-" * 30)

print("Comment Types") 
print("# This is a single-line comment") # This is also a single-line comment
print(""" This is a multi-line comment      
that spans multiple lines
and is enclosed in triple quotes """)   # This is multi-line comment   
print("-" * 30)

```

```python
# Data types
print("-" * 30)
print("Data types")
print("Integer:","42", type(42))
print("Float:","3.14", type(3.14))
print("String:","Hello, World!", type("Hello, World!"))
print("Boolean:","True", type(True))
print("List:","[1, 2, 3]", type([1, 2, 3]))
print("Tuple:","(1, 2, 3)", type((1, 2, 3)))
print("Dictionary:","{'key': 'value'}", type({'key': 'value'}))
print("Set:","{1, 2, 3}", type({1, 2, 3}))
print("-" * 30)
```

```python
# Built-in functions
print("-" * 30)
print("Built-in functions")
print("Length of 'Hello':", len("Hello"))
print("Maximum of [1, 2, 3]:", max([1, 2, 3]))
print("Minimum of [1, 2, 3]:", min([1, 2, 3]))
print("Sum of [1, 2, 3]:", sum([1, 2, 3]))
print("Type of 42:", type(42))
print("Range of 5:", list(range(5)))
print("-" * 30)

# Variable name rules
print("Variable name rules")
print("Valid variable names: my_variable, variable1, _private_var")
print("Invalid variable names: 1variable, my-variable, class")
print("-" * 30) 

# Declaring multiple variables
print("Declaring multiple variables")
a, b, c = 1, 2, 3
print("a =", a, "b =", b, "c =", c)
print("-" * 30)

# Data types and checking types
print("Data types and checking types")
x = 42
y = 3.14
z = "Hello"
print("x =", x, "type:", type(x))
print("y =", y, "type:", type(y))
print("z =", z, "type:", type(z))
print("-" * 30)
```

```python
# Assignment operators
print('-' * 30)
print("Assignment operators")
x = 10
print("Initial x =", x)
x += 5
print("After x += 5, x =", x)
x -= 3
print("After x -= 3, x =", x)
x *= 2
print("After x *= 2, x =", x)
x /= 4
print("After x /= 4, x =", x)
x //= 2
print("After x //= 2, x =", x)
x %= 3
print("After x %= 3, x =", x)
x **= 2
print("After x **= 2, x =", x)
print("-" * 30) 

# Comparison operators
print("Comparison operators")
a = 5
b = 3       
print("a =", a, "b =", b)
print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)
print("-" * 30)

# Logical operators
print("Logical operators")
x = True
y = False
print("x =", x, "y =", y)
print("x and y:", x and y)
print("x or y:", x or y)
print("not x:", not x)
print("not y:", not y)
print("-" * 30) 
```

```python
# String operations
print("-" * 30) 
print("String operations")
s1 = "Hello"
s2 = "World"
print("s1 =", s1, "s2 =", s2)
print("Concatenation: s1 + s2 =", s1 + " " + s2)
print("Repetition: s1 * 3 =", s1 * 3)
print("Length of s1:", len(s1))
print("s1 in s2:", s1 in s2)
print("s1[0]:", s1[0])
print("s1[1:4]:", s1[1:4])
print("-" * 30) 

# Escape Sequences
print("Escape sequences")
print("Newline: Hello\nWorld")
print("Tab: Hello\tWorld")
print("Backslash: Hello\\World")
print("Single quote: Hello\'s World")
print("Double quote: Hello\"s World")
print("-" * 30)

# String Formatting
print("String formatting")
name = "Alice"
age = 30
print("Using str.format(): My name is {} and I am {} years old.".format(name, age))
print(f"Using f-strings: My name is {name} and I am {age} years old.")
print("-" * 30)
```

```python
# Accessing characters in a string  
print("Accessing characters in a string")
s = "Hello, World!"
print("s =", s)
print("s[0]:", s[0])
print("s[7]:", s[7])
print("s[-1]:", s[-1])
print("s[0:5]:", s[0:5])
print("s[7:]:", s[7:])
print("s[:5]:", s[:5])
print("s[::2]:", s[::2])
print("-" * 30)

# String Methods
print("String methods")
s = "hello, world!"
print("s =", s)
print("s.upper():", s.upper()) # Converts all characters in the string to uppercase.
print("s.lower():", s.lower()) # Converts all characters in the string to lowercase.
print("s.title():", s.title()) # Converts the first character of each word to uppercase.
print("s.strip():", s.strip()) # Removes leading and trailing whitespace.
print("s.replace('o', '0'):", s.replace('o', '0')) # Replaces all occurrences of a substring with another substring.
print("s.split(', '):", s.split(', ')) # Splits the string into a list of substrings based on a delimiter.
print("s.find('world'):", s.find('world')) # Returns the lowest index of the substring if it is found in the string, otherwise returns -1.
print("s.startswith('hello'):", s.startswith('hello')) # Returns True if the string starts with the specified prefix, otherwise returns False.
print("s.endswith('!'):", s.endswith('!')) # Returns True if the string ends with the specified suffix, otherwise returns False.
print("-" * 30)

# DAY 5 List operations
```
